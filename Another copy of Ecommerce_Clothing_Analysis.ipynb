```python
import pandas as pd

data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005],
    'Product': ['T-Shirt', 'Handbag', 'Jeans', 'Dress', 'Watch'],
    'Category': ['Clothes', 'Accessories', 'Clothes', 'Clothes', 'Accessories'],
    'Price': [250, 600, 450, 700, 350],
    'Quantity': [2, 1, 1, 2, 3]
}

df = pd.DataFrame(data)

df
```

```python
df['Revenue'] = df['Price'] * df['Quantity']

df
```

```python
category_revenue = df.groupby('Category')['Revenue'].sum()

category_percentage = (category_revenue / df['Revenue'].sum()) * 100

category_percentage
```

```python
import matplotlib.pyplot as plt

category_revenue.plot(kind='bar')

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue (EGP)')
plt.show()
```

```python
product_sales = df.groupby('Product')['Quantity'].sum()

product_sales.sort_values(ascending=False)
```

```python
product_sales.sort_values(ascending=False).plot(kind='bar')

plt.title('Units Sold by Product')
plt.xlabel('Product')
plt.ylabel('Quantity Sold')
plt.show()
```

```python
data_cost = [120, 350, 250, 400, 180]

df['Cost'] = data_cost

df
```

```python
df['Total_Cost'] = df['Cost'] * df['Quantity']

df['Profit'] = df['Revenue'] - df['Total_Cost']

df
```

```python
total_profit = df['Profit'].sum()

print("Total Profit:", total_profit, "EGP")
```

```python
df[['Product', 'Revenue', 'Total_Cost', 'Profit']]
```

```python
total_revenue = df['Revenue'].sum()
total_profit = df['Profit'].sum()

profit_margin = (total_profit / total_revenue) * 100

print("Profit Margin:", profit_margin, "%")
```

```python
df['Profit_Margin'] = (df['Profit'] / df['Revenue']) * 100

df[['Product', 'Profit', 'Profit_Margin']]
```

```python
df.sort_values(
    by='Profit_Margin',
    ascending=False
)[['Product', 'Profit_Margin']]
```

```python
best_product = df.loc[df['Profit'].idxmax()]

print("Best Product:", best_product['Product'])
print("Highest Profit:", best_product['Profit'])
```

```python
df['Order_Date'] = [
    '2026-01-05',
    '2026-01-12',
    '2026-02-08',
    '2026-02-15',
    '2026-03-03'
]

df['Order_Date'] = pd.to_datetime(df['Order_Date'])

df
```

```python
df['Month'] = df['Order_Date'].dt.month_name()

df[['Order_Date', 'Month', 'Revenue']]
```

```python
monthly_sales = df.groupby('Month')['Revenue'].sum()

monthly_sales
```

```python
import numpy as np

np.random.seed(42)

print("Ready to create our dataset!")
```

```python
products = [
    'T-Shirt',
    'Jeans',
    'Dress',
    'Handbag',
    'Watch',
    'Shoes',
    'Jacket',
    'Scarf'
]

categories = [
    'Clothes',
    'Clothes',
    'Clothes',
    'Accessories',
    'Accessories',
    'Shoes',
    'Clothes',
    'Accessories'
]

print(products)
```

```python
n = 500

order_ids = np.arange(1001, 1001 + n)

product_choices = np.random.choice(products, n)

print("Number of Orders:", len(order_ids))
print("First 10 Products:", product_choices[:10])
```

```python
price_dict = {
    'T-Shirt': 250,
    'Jeans': 450,
    'Dress': 700,
    'Handbag': 600,
    'Watch': 350,
    'Shoes': 800,
    'Jacket': 900,
    'Scarf': 150
}

prices = [price_dict[p] for p in product_choices]

quantities = np.random.randint(1, 5, n)

print("First 10 Prices:", prices[:10])
print("First 10 Quantities:", quantities[:10])
```

```python
dates = pd.to_datetime(
    np.random.choice(
        pd.date_range('2026-01-01', '2026-12-31'),
        n
    )
)

print(dates[:10])
```

```python
customer_ids = np.random.choice(
    [f'C{i}' for i in range(101, 251)],
    n
)

print("First 10 Customers:", customer_ids[:10])
print("Unique Customers:", len(set(customer_ids)))
```

```python
revenues = np.array(prices) * quantities

print("First 10 Revenues:", revenues[:10])
print("Total Revenue:", revenues.sum(), "EGP")
```

```python
df = pd.DataFrame({
    'Order_ID': order_ids,
    'Customer_ID': customer_ids,
    'Product': product_choices,
    'Price': prices,
    'Quantity': quantities,
    'Order_Date': dates,
    'Revenue': revenues
})

df.head()
```

```python
category_dict = dict(zip(products, categories))

df['Category'] = df['Product'].map(category_dict)

df.head()
```

```python
cost_dict = {
    'T-Shirt': 120,
    'Jeans': 250,
    'Dress': 400,
    'Handbag': 350,
    'Watch': 180,
    'Shoes': 450,
    'Jacket': 550,
    'Scarf': 70
}

df['Cost'] = df['Product'].map(cost_dict)

df.head()
```

```python
df['Total_Cost'] = df['Cost'] * df['Quantity']

df.head()
```

```python
df['Profit'] = df['Revenue'] - df['Total_Cost']

df.head()
```
